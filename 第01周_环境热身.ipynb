## 1. 第一个 DataFrame

DataFrame 就是「一张表」，和你 Excel 里的表几乎一样。
下面手动造一张 5 行的小表，然后看看它的结构。

```python
df = pd.DataFrame({
    "城市": ["合肥", "南京", "上海", "杭州", "北京"],
    "人口_万": [963, 949, 2487, 1237, 2185],
    "GDP_亿": [12673, 16907, 44652, 18753, 41610],
})
df
```

### 常用三连：看形状、看统计、看缺失

这三个动作以后每拿到一份新数据都要先做一遍。

```python
print("形状（行, 列）：", df.shape)
print()
print("数值列的统计：")
display(df.describe())
print("缺失值：")
print(df.isna().sum())
```

## 2. 画一张中文图

matplotlib 默认不支持中文，会显示成方块。
我们已经在环境里指定了「微软雅黑」，所以下面这张图的中文是正常的。

```python
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["Microsoft YaHei"]
plt.rcParams["axes.unicode_minus"] = False

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.bar(df["城市"], df["GDP_亿"], color="#2E75B6")
ax.set_title("各城市 GDP（示例数据）")
ax.set_ylabel("亿元")
for i, v in enumerate(df["GDP_亿"]):
    ax.text(i, v + 500, str(v), ha="center", fontsize=9)
plt.tight_layout()
plt.savefig("图/示例_城市GDP.png", dpi=150)
plt.show()
print("图已保存到 图/示例_城市GDP.png")
```

## 3. 存文件 & 读回来

真实比赛里，你永远在做这两件事：**把处理结果存成 csv**、**把 csv 读回来检查**。
存的时候用 `utf-8-sig`，这样用 Excel 双击打开不会乱码。

```python
df.to_csv("数据/示例_城市.csv", index=False, encoding="utf-8-sig")
print("已写入 数据/示例_城市.csv")

back = pd.read_csv("数据/示例_城市.csv")
print("读回来的形状：", back.shape)
back
```

## 5. 报错了怎么办

1. **先看最后一行的报错**：`ModuleNotFoundError` = 包没装；`FileNotFoundError` = 路径不对。
2. **重启内核**：菜单 `Kernel → Restart Kernel`，然后从头再跑一遍。
3. **不要慌**：报错是常态，读懂第一行和最后一行就能解决 80% 的问题。
